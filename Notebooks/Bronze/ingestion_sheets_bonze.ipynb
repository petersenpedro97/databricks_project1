```python
from pyspark.sql.functions import current_timestamp, from_utc_timestamp
```

Criação e ingestão de dados da tabela de categorias

```python
df_categories = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("sep", ";")
    .option("inferSchema", "true")
    .load("/Workspace/Repos - Projetos/databricks_project1/Databases/categories.csv")
)

df_categories.printSchema()
```

```python
df_categories = df_categories.withColumn(
    "dt_ingestion",
    from_utc_timestamp(current_timestamp(), "America/Sao_Paulo")
)
```

```python
df_categories.write.mode("overwrite").saveAsTable("categories")
```

```python
df_categories.write.format("delta").mode("overwrite").saveAsTable("project1.bronze.bronze_categories")
```
